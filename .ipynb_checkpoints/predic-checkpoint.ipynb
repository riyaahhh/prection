```python
# =========================================================
# EDA on "Daily Crop Mandi Price with Weather Data (2023-25)"
# Run this in Google Colab
# =========================================================

# --- STEP 0: Install kagglehub (one-time, Colab already has most deps) ---
!pip install -q kagglehub

# --- STEP 1: Download dataset directly (no manual upload needed) ---
import kagglehub

path = kagglehub.dataset_download("jignalgajjar/daily-crop-mandi-price-with-weather-data2023-25")
print("Dataset downloaded to:", path)

import os
print("\nFiles in dataset folder:")
for f in os.listdir(path):
    print(" -", f)

# =========================================================
# STEP 2: Load the CSV
# =========================================================
import pandas as pd

# If there are multiple files, list them above and change the filename below.
csv_file = [f for f in os.listdir(path) if f.endswith(".csv")][0]
file_path = os.path.join(path, csv_file)
print("\nLoading:", file_path)

df = pd.read_csv(file_path)

# =========================================================
# STEP 3: Basic structure
# =========================================================
print("\n=== SHAPE ===")
print(df.shape)

print("\n=== COLUMNS ===")
print(df.columns.tolist())

print("\n=== DTYPES ===")
print(df.dtypes)

print("\n=== FIRST 5 ROWS ===")
print(df.head())

# =========================================================
# STEP 4: Missing values
# =========================================================
print("\n=== MISSING VALUES (count & %) ===")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_summary = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
print(missing_summary[missing_summary["missing_count"] > 0].sort_values("missing_pct", ascending=False))

# =========================================================
# STEP 5: Check for the columns we actually need
# (date, state, market, commodity, price, market_arrivals, rainfall, temperature)
# =========================================================
needed_keywords = {
    "date": ["date"],
    "state": ["state"],
    "market": ["market"],
    "commodity": ["commodity", "crop"],
    "price": ["price"],
    "arrivals": ["arrival", "quantity", "qtl", "tonnes", "volume"],
    "rainfall": ["rain"],
    "temperature": ["temp"],
    "humidity": ["humid"],
}

print("\n=== FEATURE COVERAGE CHECK ===")
cols_lower = [c.lower() for c in df.columns]
for feature, keywords in needed_keywords.items():
    matches = [df.columns[i] for i, c in enumerate(cols_lower) if any(k in c for k in keywords)]
    status = "FOUND -> " + str(matches) if matches else "MISSING"
    print(f"{feature:12s}: {status}")

# =========================================================
# STEP 6: Commodity coverage (do we have Onion, Tur/Arhar, Potato?)
# =========================================================
commodity_col = [c for c in df.columns if "commodity" in c.lower() or "crop" in c.lower()]
if commodity_col:
    col = commodity_col[0]
    print(f"\n=== UNIQUE VALUES IN '{col}' ===")
    print(df[col].value_counts())

    target_crops = ["onion", "potato", "tur", "arhar"]
    print("\n=== CHECKING FOR TARGET CROPS ===")
    for crop in target_crops:
        found = df[col].astype(str).str.lower().str.contains(crop).sum()
        print(f"{crop:10s}: {found} rows")
else:
    print("\nNo commodity/crop column found!")

# =========================================================
# STEP 7: Date range
# =========================================================
date_col = [c for c in df.columns if "date" in c.lower()]
if date_col:
    col = date_col[0]
    df[col] = pd.to_datetime(df[col], errors="coerce")
    print(f"\n=== DATE RANGE ('{col}') ===")
    print("Min date:", df[col].min())
    print("Max date:", df[col].max())
    print("Invalid/unparseable dates:", df[col].isnull().sum())

# =========================================================
# STEP 8: State / Market coverage
# =========================================================
state_col = [c for c in df.columns if "state" in c.lower()]
market_col = [c for c in df.columns if "market" in c.lower()]

if state_col:
    print(f"\n=== TOP 10 STATES ({state_col[0]}) ===")
    print(df[state_col[0]].value_counts().head(10))

if market_col:
    print(f"\n=== NUMBER OF UNIQUE MARKETS ({market_col[0]}) ===")
    print(df[market_col[0]].nunique())

# =========================================================
# STEP 9: Quick numeric summary (price, weather etc.)
# =========================================================
print("\n=== NUMERIC SUMMARY (describe) ===")
print(df.describe())

# =========================================================
# STEP 10: Save a small sample to inspect / share if needed
# =========================================================
sample = df.sample(min(2000, len(df)), random_state=42)
sample.to_csv("mandi_sample.csv", index=False)
print("\nSaved a 2000-row sample as mandi_sample.csv (in Colab's file panel) for quick inspection.")

print("\n=== EDA COMPLETE ===")
```

```python
# =========================================================
# ARIMA + XGBoost Price Forecasting Pipeline
# Onion + Potato | 7/14/30-day horizons
# Assumes `df` already loaded from the EDA script (kagglehub)
# =========================================================

!pip install -q pmdarima xgboost

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

# =========================================================
# STEP 1: Filter to Onion + Potato only, clean dates
# =========================================================
df["date"] = pd.to_datetime(df["date"])
df = df[df["Commodity"].isin(["Onion", "Potato"])].copy()
df = df.sort_values(["Commodity", "STATE", "Market Name", "date"])

print("Filtered shape:", df.shape)
print(df["Commodity"].value_counts())

# =========================================================
# STEP 2: Pick ONE market per commodity to start with
# (time-constrained MVP — expand to more markets later once this works)
# =========================================================
top_market_onion = (
    df[df["Commodity"] == "Onion"]["Market Name"].value_counts().idxmax()
)
top_market_potato = (
    df[df["Commodity"] == "Potato"]["Market Name"].value_counts().idxmax()
)
print("Onion market chosen:", top_market_onion)
print("Potato market chosen:", top_market_potato)

def get_series(commodity, market):
    sub = df[(df["Commodity"] == commodity) & (df["Market Name"] == market)].copy()
    sub = sub.drop_duplicates(subset="date").sort_values("date")
    sub = sub.set_index("date").asfreq("D")  # force daily frequency
    sub["Modal_Price"] = sub["Modal_Price"].interpolate(limit=5)  # fill small gaps
    return sub.dropna(subset=["Modal_Price"])

onion_df = get_series("Onion", top_market_onion)
potato_df = get_series("Potato", top_market_potato)

print("\nOnion series shape:", onion_df.shape)
print("Potato series shape:", potato_df.shape)

# =========================================================
# STEP 3: Feature engineering (most already exists — add what's missing)
# =========================================================
def add_features(data):
    d = data.copy()
    d["month"] = d.index.month
    d["day_of_year"] = d.index.dayofyear
    d["volatility_7d"] = d["Modal_Price"].rolling(7).std()
    d["price_change_1d"] = d["Modal_Price"].pct_change(1)
    return d

onion_df = add_features(onion_df)
potato_df = add_features(potato_df)

# =========================================================
# STEP 4: Time-based train/test split (NEVER random split for time series)
# =========================================================
def time_split(data, test_days=60):
    split_point = data.index.max() - pd.Timedelta(days=test_days)
    train = data[data.index <= split_point]
    test = data[data.index > split_point]
    return train, test

onion_train, onion_test = time_split(onion_df)
potato_train, potato_test = time_split(potato_df)

print("\nOnion train/test:", onion_train.shape, onion_test.shape)
print("Potato train/test:", potato_train.shape, potato_test.shape)

# =========================================================
# STEP 5: ARIMA baseline (univariate, price-only)
# =========================================================
from pmdarima import auto_arima
from sklearn.metrics import mean_absolute_error, mean_squared_error

def run_arima(train, test, name):
    print(f"\n--- ARIMA: {name} ---")
    model = auto_arima(
        train["Modal_Price"],
        seasonal=False,
        stepwise=True,
        suppress_warnings=True,
        error_action="ignore",
        trace=False,
    )
    forecast = model.predict(n_periods=len(test))
    mae = mean_absolute_error(test["Modal_Price"], forecast)
    rmse = np.sqrt(mean_squared_error(test["Modal_Price"], forecast))

    # Safe MAPE: skip rows where actual price is 0 (avoids div-by-zero -> NaN)
    actual = test["Modal_Price"].values
    nonzero_mask = actual != 0
    if nonzero_mask.sum() == 0:
        mape = np.nan
        print(f"  Warning: all actual prices are 0 in test set for {name} — MAPE undefined")
    else:
        mape = np.mean(np.abs((actual[nonzero_mask] - forecast[nonzero_mask]) / actual[nonzero_mask])) * 100
        zero_count = (~nonzero_mask).sum()
        if zero_count > 0:
            print(f"  Note: {zero_count} row(s) with actual price = 0 excluded from MAPE calc")

    print(f"MAE: {mae:.2f} | RMSE: {rmse:.2f} | MAPE: {mape:.2f}%")
    return model, forecast, {"MAE": mae, "RMSE": rmse, "MAPE": mape}

onion_arima_model, onion_arima_forecast, onion_arima_metrics = run_arima(onion_train, onion_test, "Onion")
potato_arima_model, potato_arima_forecast, potato_arima_metrics = run_arima(potato_train, potato_test, "Potato")

# =========================================================
# STEP 6: XGBoost multi-horizon (7, 14, 30 days ahead)
# =========================================================
from xgboost import XGBRegressor

FEATURE_COLS = [
    "Modal_Price", "temp_mean", "rainfall_mm", "humidity",
    "price_lag_1d", "price_lag_7d", "price_lag_14d", "price_lag_30d",
    "price_7d_avg", "price_30d_avg",
    "rainfall_mm_7d_avg", "rainfall_mm_30d_avg",
    "temp_mean_7d_avg", "temp_mean_30d_avg",
    "volatility_7d", "price_change_1d",
    "month", "day_of_year",
]

def make_targets(data, horizons=[7, 14, 30]):
    d = data.copy()
    for h in horizons:
        d[f"target_{h}d"] = d["Modal_Price"].shift(-h)
    return d

def train_xgb_multi_horizon(train, test, name, horizons=[7, 14, 30]):
    print(f"\n--- XGBoost: {name} ---")
    train_full = make_targets(train, horizons)
    test_full = make_targets(test, horizons)

    results = {}
    models = {}

    for h in horizons:
        cols = [c for c in FEATURE_COLS if c in train_full.columns]
        X_train = train_full[cols].copy()
        y_train = train_full[f"target_{h}d"]

        X_test = test_full[cols].copy()
        y_test = test_full[f"target_{h}d"]

        # drop rows where target is NaN (end of series, shift artifact)
        train_mask = y_train.notna() & X_train.notna().all(axis=1)
        test_mask = y_test.notna() & X_test.notna().all(axis=1)

        X_train_clean, y_train_clean = X_train[train_mask], y_train[train_mask]
        X_test_clean, y_test_clean = X_test[test_mask], y_test[test_mask]

        if len(X_test_clean) == 0:
            print(f"  [{h}d] Not enough test data after cleaning — skipping")
            continue

        model = XGBRegressor(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42,
        )
        model.fit(X_train_clean, y_train_clean)
        preds = model.predict(X_test_clean)

        mae = mean_absolute_error(y_test_clean, preds)
        rmse = np.sqrt(mean_squared_error(y_test_clean, preds))
        mape = np.mean(np.abs((y_test_clean - preds) / y_test_clean)) * 100

        print(f"  [{h}d] MAE: {mae:.2f} | RMSE: {rmse:.2f} | MAPE: {mape:.2f}%")

        results[h] = {"MAE": mae, "RMSE": rmse, "MAPE": mape}
        models[h] = model

    return models, results

onion_xgb_models, onion_xgb_results = train_xgb_multi_horizon(onion_train, onion_test, "Onion")
potato_xgb_models, potato_xgb_results = train_xgb_multi_horizon(potato_train, potato_test, "Potato")

# =========================================================
# STEP 7: Comparison table — ARIMA vs XGBoost
# =========================================================
print("\n" + "=" * 60)
print("MODEL COMPARISON SUMMARY")
print("=" * 60)

comparison_rows = []
for commodity, arima_metrics, xgb_results in [
    ("Onion", onion_arima_metrics, onion_xgb_results),
    ("Potato", potato_arima_metrics, potato_xgb_results),
]:
    comparison_rows.append({
        "Commodity": commodity, "Model": "ARIMA (baseline)", "Horizon": "n/a",
        "MAE": round(arima_metrics["MAE"], 2), "RMSE": round(arima_metrics["RMSE"], 2),
        "MAPE": round(arima_metrics["MAPE"], 2),
    })
    for h, m in xgb_results.items():
        comparison_rows.append({
            "Commodity": commodity, "Model": "XGBoost", "Horizon": f"{h}d",
            "MAE": round(m["MAE"], 2), "RMSE": round(m["RMSE"], 2), "MAPE": round(m["MAPE"], 2),
        })

comparison_df = pd.DataFrame(comparison_rows)
print(comparison_df.to_string(index=False))

comparison_df.to_csv("model_comparison_results.csv", index=False)
print("\nSaved comparison table as model_comparison_results.csv")

# =========================================================
# STEP 8: Save trained models for later use (dashboard/SHAP)
# =========================================================
import joblib

for h, model in onion_xgb_models.items():
    joblib.dump(model, f"onion_xgb_{h}d.pkl")
for h, model in potato_xgb_models.items():
    joblib.dump(model, f"potato_xgb_{h}d.pkl")

joblib.dump(onion_arima_model, "onion_arima.pkl")
joblib.dump(potato_arima_model, "potato_arima.pkl")

print("\nAll models saved. Check Colab's file panel.")
print("\n=== TRAINING COMPLETE ===")
```

```python
# =========================================================
# Forecast + Risk + Reasoning + Buffer Stock Engine
# Run this AFTER your training script (needs onion_df, potato_df,
# onion_xgb_models, potato_xgb_models already in memory)
# =========================================================

!pip install -q shap

import shap
import numpy as np
import pandas as pd

FEATURE_COLS = [
    "Modal_Price", "temp_mean", "rainfall_mm", "humidity",
    "price_lag_1d", "price_lag_7d", "price_lag_14d", "price_lag_30d",
    "price_7d_avg", "price_30d_avg",
    "rainfall_mm_7d_avg", "rainfall_mm_30d_avg",
    "temp_mean_7d_avg", "temp_mean_30d_avg",
    "volatility_7d", "price_change_1d",
    "month", "day_of_year",
]

# Human-readable labels for SHAP reasoning output
FEATURE_LABELS = {
    "Modal_Price": "current price level",
    "temp_mean": "current temperature",
    "rainfall_mm": "recent rainfall",
    "humidity": "humidity levels",
    "price_lag_1d": "yesterday's price",
    "price_lag_7d": "price a week ago",
    "price_lag_14d": "price two weeks ago",
    "price_lag_30d": "price a month ago",
    "price_7d_avg": "7-day average price trend",
    "price_30d_avg": "30-day average price trend",
    "rainfall_mm_7d_avg": "recent weekly rainfall pattern",
    "rainfall_mm_30d_avg": "recent monthly rainfall pattern",
    "temp_mean_7d_avg": "weekly temperature trend",
    "temp_mean_30d_avg": "monthly temperature trend",
    "volatility_7d": "recent price volatility",
    "price_change_1d": "yesterday's price movement",
    "month": "seasonal/monthly pattern",
    "day_of_year": "time-of-year effect",
}

COMMODITY_CONFIG = {
    "Onion": {"df": None, "models": None},   # filled in below
    "Potato": {"df": None, "models": None},
}

# Wire in your already-trained objects
COMMODITY_CONFIG["Onion"]["df"] = onion_df
COMMODITY_CONFIG["Onion"]["models"] = onion_xgb_models
COMMODITY_CONFIG["Potato"]["df"] = potato_df
COMMODITY_CONFIG["Potato"]["models"] = potato_xgb_models


def get_latest_feature_row(data):
    """Grab the most recent row with complete features to predict from 'today'."""
    cols = [c for c in FEATURE_COLS if c in data.columns]
    clean = data.dropna(subset=cols)
    latest_row = clean.iloc[[-1]]  # keep as DataFrame, not Series
    return latest_row[cols], latest_row.index[0]


# Group raw model features into human-readable categories for reasoning output
FEATURE_CATEGORIES = {
    "Price momentum": [
        "Modal_Price", "price_lag_1d", "price_lag_7d", "price_lag_14d", "price_lag_30d",
        "price_7d_avg", "price_30d_avg",
    ],
    # NOTE: We don't have real market_arrivals data. Price volatility and
    # sharp short-term price moves are a real, defensible PROXY for supply
    # pressure — arrivals shortages cause spikes, arrival gluts cause drops.
    # This is genuine SHAP-derived signal, just honestly relabeled — not
    # fabricated data.
    "Supply pressure (proxy)": ["volatility_7d", "price_change_1d"],
    "Rainfall": ["rainfall_mm", "rainfall_mm_7d_avg", "rainfall_mm_30d_avg"],
    "Temperature & humidity": ["temp_mean", "temp_mean_7d_avg", "temp_mean_30d_avg", "humidity"],
    "Seasonality": ["month", "day_of_year"],
}

CATEGORY_NOTES = {
    "Supply pressure (proxy)": "Estimated from price volatility, since live market arrivals data wasn't available — price already reacts to supply shocks.",
}

# Categories with genuinely no data or proxy available.
UNAVAILABLE_CATEGORIES = ["Transport risk"]


def explain_prediction(model, X_row, current_price, predicted_price):
    """
    Use SHAP to explain what's driving this prediction, grouped into
    named categories (Price momentum, Rainfall, etc.) instead of raw
    feature names — matches the dashboard-style breakdown format.
    """
    explainer = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_row)[0]
    feature_shap = dict(zip(X_row.columns, shap_values))

    total_abs_impact = float(np.sum(np.abs(shap_values)))

    category_results = []
    for category, features in FEATURE_CATEGORIES.items():
        cat_shap_sum = sum(abs(feature_shap.get(f, 0)) for f in features)
        pct_impact = (cat_shap_sum / total_abs_impact * 100) if total_abs_impact > 0 else 0
        entry = {
            "category": category,
            "impact_pct": round(float(pct_impact), 1),
        }
        if category in CATEGORY_NOTES:
            entry["note"] = CATEGORY_NOTES[category]
        category_results.append(entry)

    # Add genuinely unavailable categories at null — no fabricated data
    for category in UNAVAILABLE_CATEGORIES:
        category_results.append({
            "category": category,
            "impact_pct": None,
            "note": "Not available — no data source or reliable proxy for this in the current dataset",
        })

    # Sort available categories by impact, unavailable ones stay at the end
    category_results.sort(key=lambda x: (x["impact_pct"] is None, -(x["impact_pct"] or 0)))

    # Net summary — always anchored to TODAY'S actual price
    pct_change = ((predicted_price - current_price) / current_price) * 100
    move_word = "rise" if pct_change > 0 else "fall"
    top_category = next((c["category"] for c in category_results if c["impact_pct"] is not None), "recent trends")

    net_summary = (
        f"Price is forecast to {move_word} {abs(pct_change):.1f}% "
        f"(₹{current_price:.0f} → ₹{predicted_price:.0f}), "
        f"most strongly influenced by {top_category.lower()}."
    )

    return {"net_summary": net_summary, "factor_breakdown": category_results}


def assess_risk(current_price, forecast_7d, forecast_14d, forecast_30d, volatility):
    """Rule-based risk classification."""
    pct_change_7d = ((forecast_7d - current_price) / current_price) * 100
    pct_change_30d = ((forecast_30d - current_price) / current_price) * 100

    # Thresholds — tune these based on what looks reasonable for your commodities
    high_risk = (abs(pct_change_7d) > 15 or abs(pct_change_30d) > 25) and volatility > 50
    watch = (abs(pct_change_7d) > 8 or abs(pct_change_30d) > 15)

    if high_risk:
        level = "High Risk"
    elif watch:
        level = "Watch"
    else:
        level = "Stable"

    return level, pct_change_7d, pct_change_30d


def buffer_stock_recommendation(risk_level, pct_change_7d, commodity):
    """
    Rule-based decision support with a quantified estimate.

    NOTE: We don't have real government buffer stock or market arrivals
    data. To make the recommendation concrete instead of just a label,
    we assume a benchmark average daily arrival volume per market
    (a reasonable illustrative placeholder, not real data) and scale
    the suggested action off the forecasted price gap. This is clearly
    a simulated decision-support layer, not a live policy feed —
    fine to say so directly if asked.
    """
    # Assumed benchmark daily arrivals per market (quintals) — placeholder,
    # since real market_arrivals data wasn't available for this MVP.
    ASSUMED_DAILY_ARRIVALS = {"Onion": 500, "Potato": 600}
    baseline_arrivals = ASSUMED_DAILY_ARRIVALS.get(commodity, 500)

    # Intervention scaling factor: how much of the price gap we try to
    # offset via supply adjustment. 0.5 is a conservative illustrative
    # assumption, not a calibrated economic parameter.
    INTERVENTION_FACTOR = 0.5

    if risk_level == "High Risk" and pct_change_7d > 0:
        qty = round(baseline_arrivals * (pct_change_7d / 100) * INTERVENTION_FACTOR)
        return (
            f"Recommend releasing approximately {qty} quintals of buffer stock "
            f"over the next 7 days to help offset the forecasted {pct_change_7d:.1f}% price rise "
            f"(estimate based on an assumed average arrival volume of {baseline_arrivals} quintals/day — "
            f"illustrative, not live government data)."
        )
    elif risk_level == "High Risk" and pct_change_7d < 0:
        qty = round(baseline_arrivals * (abs(pct_change_7d) / 100) * INTERVENTION_FACTOR)
        return (
            f"Recommend procuring approximately {qty} quintals to support farmer prices "
            f"given the forecasted {abs(pct_change_7d):.1f}% price drop "
            f"(estimate based on an assumed average arrival volume of {baseline_arrivals} quintals/day — "
            f"illustrative, not live government data)."
        )
    elif risk_level == "Watch":
        return "Monitor closely over the next week — no stock intervention recommended yet."
    else:
        return "Market is stable. No buffer stock intervention needed."


def get_full_forecast(commodity):
    """Main function: returns the complete output block for one commodity."""
    config = COMMODITY_CONFIG[commodity]
    data = config["df"]
    models = config["models"]

    X_row, latest_date = get_latest_feature_row(data)
    current_price = float(X_row["Modal_Price"].values[0])
    volatility = float(X_row["volatility_7d"].values[0])

    forecasts = {}
    explanations = {}
    for h in [7, 14, 30]:
        model = models[h]
        pred = model.predict(X_row)[0]
        forecasts[h] = round(float(pred), 2)
        explanations[h] = explain_prediction(model, X_row, current_price, forecasts[h])

    risk_level, pct_change_7d, pct_change_30d = assess_risk(
        current_price, forecasts[7], forecasts[14], forecasts[30], volatility
    )
    buffer_action = buffer_stock_recommendation(risk_level, pct_change_7d, commodity)

    result = {
        "commodity": commodity,
        "as_of_date": str(latest_date.date()),
        "current_price": round(float(current_price), 2),
        "forecast_7d": forecasts[7],
        "forecast_14d": forecasts[14],
        "forecast_30d": forecasts[30],
        "pct_change_7d": round(pct_change_7d, 2),
        "pct_change_30d": round(pct_change_30d, 2),
        "risk_level": risk_level,
        "explanation_7d": explanations[7],
        "explanation_30d": explanations[30],
        "buffer_stock_recommendation": buffer_action,
    }
    return result


# =========================================================
# RUN IT — this is what your API endpoint will eventually return
# =========================================================
import json

print("=" * 60)
print("ONION FORECAST")
print("=" * 60)
onion_result = get_full_forecast("Onion")
print(json.dumps(onion_result, indent=2))

print("\n" + "=" * 60)
print("POTATO FORECAST")
print("=" * 60)
potato_result = get_full_forecast("Potato")
print(json.dumps(potato_result, indent=2))
```
